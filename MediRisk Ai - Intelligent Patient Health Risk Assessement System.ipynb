```python
# ==========================================================
#                 MEDIRISK AI(1.0)
#      Intelligent Patient Health Risk Assessment
# ==========================================================

# Import Libraries

from datetime import datetime
import csv
import os


# Hospital Details

hospital_name = "City Care Hospital"
department = "General Medicine"
doctor_name = "Dr. Kumkum"

# Date and Time

today = datetime.now()

current_date = today.strftime("%d-%m-%Y")
current_time = today.strftime("%I:%M %p")


# Welcome Screen

print("=" * 60)
print("                MEDIRISK AI")
print("    Intelligent Patient Health Assessment")
print("=" * 60)

print(f"Hospital   : {hospital_name}")
print(f"Department : {department}")
print(f"Doctor     : {doctor_name}")
print(f"Date       : {current_date}")
print(f"Time       : {current_time}")

print("=" * 60)


# ==========================================================
# BMI Calculator
# ==========================================================

def bmi(weight, height):

    value = weight / (height ** 2)

    return round(value, 2)


# ==========================================================
# BMI Status
# ==========================================================

def bmi_status(value):

    if value < 18.5:
        return "Underweight"

    elif value < 25:
        return "Normal"

    elif value < 30:
        return "Overweight"

    else:
        return "Obese"


# ==========================================================
# Age Category
# ==========================================================

def age_category(age):

    if age < 18:
        return "Child"

    elif age < 40:
        return "Young Adult"

    elif age < 60:
        return "Adult"

    else:
        return "Senior Citizen"


# ==========================================================
# Body Temperature
# ==========================================================

def temp_status(temp):

    if temp < 37.5:
        return "Normal"

    elif temp <= 38.5:
        return "Fever"

    else:
        return "High Fever"


# ==========================================================
# Oxygen Level
# ==========================================================

def oxygen_status(oxygen):

    if oxygen >= 95:
        return "Normal"

    elif oxygen >= 90:
        return "Low"

    else:
        return "Critical"


# ==========================================================
# Heart Rate
# ==========================================================

def heart_status(rate):

    if 60 <= rate <= 100:
        return "Normal"

    elif rate > 100:
        return "High"

    else:
        return "Low"


# ==========================================================
# Blood Pressure
# ==========================================================

def bp_status(bp):

    bp = bp.lower()

    if bp == "normal":
        return "Normal"

    return "High"


# ==========================================================
# Blood Sugar
# ==========================================================

def sugar_status(sugar):

    if sugar < 140:
        return "Normal"

    elif sugar < 180:
        return "Borderline"

    else:
        return "High"
# ==========================================================
# Risk Score Calculation
# ==========================================================

def calculate_risk(age, temp, oxygen, heart_rate,
                   bp, sugar, smoking,
                   diabetes, kidney, asthma):

    score = 0

    # Age Risk
    if age >= 60:
        score += 2

    # Temperature Risk
    if temp >= 38:
        score += 2

    # Oxygen Risk
    if oxygen < 95:
        score += 3

    # Heart Rate Risk
    if heart_rate < 60 or heart_rate > 100:
        score += 2

    # Blood Pressure Risk
    if bp.lower() == "high":
        score += 2

    # Sugar Risk
    if sugar >= 180:
        score += 2

    # Lifestyle Risk
    if smoking.lower() == "yes":
        score += 2

    # Medical History Risk
    if diabetes.lower() == "yes":
        score += 3

    if kidney.lower() == "yes":
        score += 3

    if asthma.lower() == "yes":
        score += 2

    return score


# ==========================================================
# Risk Level
# ==========================================================

def risk_level(score):

    if score <= 4:
        return "LOW"

    elif score <= 8:
        return "MEDIUM"

    elif score <= 12:
        return "HIGH"

    else:
        return "CRITICAL"


# ==========================================================
# Health Percentage
# ==========================================================

def health_percentage(score):

    health = 100 - (score * 5)

    if health < 0:
        health = 0

    return health


# ==========================================================
# Health Grade
# ==========================================================

def health_grade(health):

    if health >= 90:
        return "A+"

    elif health >= 80:
        return "A"

    elif health >= 70:
        return "B"

    elif health >= 60:
        return "C"

    else:
        return "D"


# ==========================================================
# Water Intake
# ==========================================================

def water_need(weight):

    amount = weight * 35

    return round(amount / 1000, 1)


# ==========================================================
# Exercise Suggestion
# ==========================================================

def exercise_plan(age):

    if age < 40:
        return "Running / Gym"

    elif age < 60:
        return "Brisk Walking"

    else:
        return "Walking and Yoga"


# ==========================================================
# Diet Suggestion
# ==========================================================

def diet_plan(diabetes):

    if diabetes.lower() == "yes":
        return "Low Sugar Diet"

    else:
        return "Balanced Diet"


# ==========================================================
# Sleep Advice
# ==========================================================

def sleep_advice(age):

    if age < 18:
        return "8-10 Hours"

    elif age < 60:
        return "7-8 Hours"

    else:
        return "7 Hours"


# ==========================================================
# Next Checkup
# ==========================================================

def checkup_time(level):

    if level == "LOW":
        return "After 6 Months"

    elif level == "MEDIUM":
        return "After 3 Months"

    elif level == "HIGH":
        return "Within 1 Week"

    else:
        return "Immediately"


# ==========================================================
# Doctor Advice
# ==========================================================

def doctor_advice(level):

    if level == "LOW":

        return "Maintain healthy lifestyle and regular exercise."

    elif level == "MEDIUM":

        return "Monitor health regularly and take precautions."

    elif level == "HIGH":

        return "Consult doctor as soon as possible."

    else:

        return "Emergency medical attention required."
# ==========================================================
# Patient Registration
# ==========================================================

print("\n")
print("=" * 60)
print("              PATIENT REGISTRATION")
print("=" * 60)


# Patient Details

patient_id = input("Enter Patient ID : ").upper()

name = input("Enter Patient Name : ").title()


# Age Input

while True:

    age = int(input("Enter Age : "))

    if age > 0 and age <= 120:
        break

    else:
        print("Please enter a valid age.")


# Gender Input

while True:

    gender = input("Enter Gender (Male/Female) : ").title()

    if gender == "Male" or gender == "Female":
        break

    else:
        print("Enter Male or Female only.")


# Height Input

while True:

    height = float(input("Enter Height in metres : "))

    if height > 0:
        break

    else:
        print("Height should be greater than 0.")


# Weight Input

while True:

    weight = float(input("Enter Weight in kg : "))

    if weight > 0:
        break

    else:
        print("Weight should be greater than 0.")



# ==========================================================
# Health Parameters
# ==========================================================


# Temperature

while True:

    temp = float(input("Enter Body Temperature : "))

    if 30 <= temp <= 45:
        break

    else:
        print("Enter valid temperature.")



# Oxygen Level

while True:

    oxygen = int(input("Enter Oxygen Level (%) : "))

    if 50 <= oxygen <= 100:
        break

    else:
        print("Oxygen should be between 50-100.")



# Heart Rate

while True:

    heart_rate = int(input("Enter Heart Rate : "))

    if 20 <= heart_rate <= 220:
        break

    else:
        print("Enter valid heart rate.")



# Blood Pressure

while True:

    bp = input("Blood Pressure (Normal/High) : ").lower()

    if bp == "normal" or bp == "high":
        break

    else:
        print("Enter only Normal or High.")



# Sugar Level

while True:

    sugar = int(input("Enter Sugar Level : "))

    if sugar > 0:
        break

    else:
        print("Enter valid sugar level.")



# ==========================================================
# Medical History
# ==========================================================


smoking = input("Smoking (Yes/No) : ").lower()

diabetes = input("Diabetes (Yes/No) : ").lower()

kidney = input("Kidney Disease (Yes/No) : ").lower()

asthma = input("Asthma (Yes/No) : ").lower()



print("\nPatient information saved successfully.")
# ==========================================================
# Health Calculation
# ==========================================================


# BMI Calculation

bmi_value = bmi(weight, height)

bmi_result = bmi_status(bmi_value)


# Age Category

age_result = age_category(age)


# Vital Signs Status

temperature_result = temp_status(temp)

oxygen_result = oxygen_status(oxygen)

heart_result = heart_status(heart_rate)

bp_result = bp_status(bp)

sugar_result = sugar_status(sugar)



# ==========================================================
# Risk Calculation
# ==========================================================

risk_score = calculate_risk(
    age,
    temp,
    oxygen,
    heart_rate,
    bp,
    sugar,
    smoking,
    diabetes,
    kidney,
    asthma
)


# Risk Level

risk = risk_level(risk_score)



# Health Percentage

health = health_percentage(risk_score)



# Health Grade

grade = health_grade(health)



# ==========================================================
# Lifestyle Suggestions
# ==========================================================

water = water_need(weight)

exercise = exercise_plan(age)

diet = diet_plan(diabetes)

sleep = sleep_advice(age)

next_checkup = checkup_time(risk)


# Doctor Advice

advice = doctor_advice(risk)



print("\nHealth calculation completed successfully.")
# ==========================================================
#                MEDIRISK AI HEALTH REPORT
# ==========================================================


print("\n")
print("=" * 65)
print("              MEDIRISK AI HEALTH REPORT")
print("=" * 65)



# Patient Details

print("\nPatient Details")
print("-" * 40)

print(f"Patient ID       : {patient_id}")
print(f"Name             : {name}")
print(f"Age              : {age} years")
print(f"Age Category     : {age_result}")
print(f"Gender           : {gender}")



# Body Details

print("\nBody Information")
print("-" * 40)

print(f"Height           : {height} m")
print(f"Weight           : {weight} kg")
print(f"BMI              : {bmi_value}")
print(f"BMI Status       : {bmi_result}")



# Vital Signs

print("\nVital Signs")
print("-" * 40)

print(f"Temperature      : {temp} °C")
print(f"Temperature      : {temperature_result}")

print(f"Oxygen Level     : {oxygen}%")
print(f"Oxygen Status    : {oxygen_result}")

print(f"Heart Rate       : {heart_rate} BPM")
print(f"Heart Status     : {heart_result}")

print(f"Blood Pressure   : {bp}")
print(f"BP Status        : {bp_result}")

print(f"Sugar Level      : {sugar} mg/dL")
print(f"Sugar Status     : {sugar_result}")



# Medical History

print("\nMedical History")
print("-" * 40)

print(f"Smoking          : {smoking.title()}")
print(f"Diabetes         : {diabetes.title()}")
print(f"Kidney Disease   : {kidney.title()}")
print(f"Asthma           : {asthma.title()}")



# Risk Analysis

print("\nRisk Analysis")
print("-" * 40)

print(f"Risk Score       : {risk_score}")
print(f"Risk Level       : {risk}")
print(f"Health Score     : {health}%")
print(f"Health Grade     : {grade}")



# Lifestyle Advice

print("\nLifestyle Advice")
print("-" * 40)

print(f"Water Intake     : {water} litres/day")
print(f"Exercise Plan    : {exercise}")
print(f"Diet Plan        : {diet}")
print(f"Sleep            : {sleep}")
print(f"Next Checkup     : {next_checkup}")



# Doctor Advice

print("\nDoctor Recommendation")
print("-" * 40)

print(advice)



print("\n" + "=" * 65)
print("          Report Generated Successfully")
print("=" * 65)
# ==========================================================
#              RISK SCORE BREAKDOWN
# ==========================================================

print("\n")
print("=" * 65)
print("              RISK SCORE BREAKDOWN")
print("=" * 65)


risk_found = False


if age >= 60:

    print("Age above 60                 : +2 Risk")
    risk_found = True


if temp >= 38:

    print("High Temperature             : +2 Risk")
    risk_found = True


if oxygen < 95:

    print("Low Oxygen Level             : +3 Risk")
    risk_found = True


if heart_rate < 60 or heart_rate > 100:

    print("Abnormal Heart Rate          : +2 Risk")
    risk_found = True


if bp.lower() == "high":

    print("High Blood Pressure           : +2 Risk")
    risk_found = True


if sugar >= 180:

    print("High Sugar Level              : +2 Risk")
    risk_found = True


if smoking == "yes":

    print("Smoking Habit                 : +2 Risk")
    risk_found = True


if diabetes == "yes":

    print("Diabetes History              : +3 Risk")
    risk_found = True


if kidney == "yes":

    print("Kidney Disease History        : +3 Risk")
    risk_found = True


if asthma == "yes":

    print("Asthma History                : +2 Risk")
    risk_found = True



if risk_found == False:

    print("No major risk factors found.")



print("-" * 65)
print(f"Total Risk Score : {risk_score}")



# ==========================================================
#              EMERGENCY ALERT
# ==========================================================

print("\n")
print("=" * 65)
print("              EMERGENCY CHECK")
print("=" * 65)


if risk == "CRITICAL" or oxygen < 90:

    print("⚠ Emergency Alert!")
    print("Immediate medical attention is required.")

elif risk == "HIGH":

    print("High Risk Patient")
    print("Doctor consultation is recommended.")

else:

    print("No emergency condition detected.")



# ==========================================================
#              PERSONAL HEALTH TIPS
# ==========================================================

print("\n")
print("=" * 65)
print("              PERSONAL HEALTH TIPS")
print("=" * 65)



if bmi_result == "Underweight":

    print("- Increase nutritious food intake.")

elif bmi_result == "Overweight":

    print("- Maintain daily exercise routine.")

elif bmi_result == "Obese":

    print("- Follow weight management plan.")

else:

    print("- Maintain your current healthy lifestyle.")



if smoking == "yes":

    print("- Try to reduce or quit smoking.")


if diabetes == "yes":

    print("- Check sugar level regularly.")


if oxygen < 95:

    print("- Monitor oxygen level frequently.")


if bp == "high":

    print("- Reduce salt intake and monitor BP.")


print("=" * 65)
# ==========================================================
#              SAVE PATIENT REPORT (TXT FILE)
# ==========================================================


report_file = open("Patient_Report.txt", "w")


report_file.write("MEDIRISK AI HEALTH REPORT\n")
report_file.write("=" * 50 + "\n\n")


report_file.write("Patient Details\n")
report_file.write("-" * 30 + "\n")

report_file.write(f"Patient ID : {patient_id}\n")
report_file.write(f"Name       : {name}\n")
report_file.write(f"Age        : {age}\n")
report_file.write(f"Gender     : {gender}\n\n")


report_file.write("Health Details\n")
report_file.write("-" * 30 + "\n")

report_file.write(f"BMI        : {bmi_value}\n")
report_file.write(f"BMI Status : {bmi_result}\n")

report_file.write(f"Risk Score : {risk_score}\n")
report_file.write(f"Risk Level : {risk}\n")

report_file.write(f"Health     : {health}%\n")
report_file.write(f"Grade      : {grade}\n\n")


report_file.write("Doctor Advice\n")
report_file.write("-" * 30 + "\n")

report_file.write(advice)


report_file.close()


print("\nPatient report saved successfully.")



# ==========================================================
#              SAVE DATA IN CSV FILE
# ==========================================================


file_exists = os.path.isfile("patient_data.csv")


with open("patient_data.csv", "a", newline="") as file:

    writer = csv.writer(file)


    if file_exists == False:

        writer.writerow([
            "Patient ID",
            "Name",
            "Age",
            "Gender",
            "BMI",
            "Risk Score",
            "Risk Level",
            "Health Score"
        ])



    writer.writerow([
        patient_id,
        name,
        age,
        gender,
        bmi_value,
        risk_score,
        risk,
        health
    ])



print("Patient data saved successfully.")
# ==========================================================
#             MULTIPLE PATIENT SUMMARY
# ==========================================================


patients = []


# Current patient data store

patient_info = {

    "id": patient_id,
    "name": name,
    "age": age,
    "bmi": bmi_value,
    "risk_score": risk_score,
    "risk": risk,
    "health": health

}


patients.append(patient_info)



# ==========================================================
#             HOSPITAL SUMMARY
# ==========================================================


print("\n")
print("=" * 65)
print("              HOSPITAL SUMMARY")
print("=" * 65)



total_patient = len(patients)

print(f"Total Patients : {total_patient}")



low = 0
medium = 0
high = 0
critical = 0



for patient in patients:


    if patient["risk"] == "LOW":

        low += 1


    elif patient["risk"] == "MEDIUM":

        medium += 1


    elif patient["risk"] == "HIGH":

        high += 1


    else:

        critical += 1



print("\nRisk Category Count")
print("-" * 40)

print(f"Low Risk       : {low}")
print(f"Medium Risk    : {medium}")
print(f"High Risk      : {high}")
print(f"Critical Risk  : {critical}")



# ==========================================================
#             AVERAGE BMI
# ==========================================================


total_bmi = 0


for patient in patients:

    total_bmi += patient["bmi"]



average_bmi = total_bmi / total_patient


print("\nAverage BMI")
print("-" * 40)

print(round(average_bmi, 2))



# ==========================================================
#             HIGHEST RISK PATIENT
# ==========================================================


highest_risk = patients[0]


for patient in patients:


    if patient["risk_score"] > highest_risk["risk_score"]:

        highest_risk = patient



print("\nHighest Risk Patient")
print("-" * 40)

print(f"Name        : {highest_risk['name']}")
print(f"Risk Score  : {highest_risk['risk_score']}")
print(f"Risk Level  : {highest_risk['risk']}")



print("=" * 65)
# ==========================================================
#             PATIENT MANAGEMENT SYSTEM
# ==========================================================


def search_patient(patient_list):

    search_id = input("Enter Patient ID to search : ").upper()


    found = False


    for patient in patient_list:


        if patient["id"] == search_id:

            print("\nPatient Found")
            print("-" * 40)

            print(f"Name        : {patient['name']}")
            print(f"Age         : {patient['age']}")
            print(f"BMI         : {patient['bmi']}")
            print(f"Risk Score  : {patient['risk_score']}")
            print(f"Risk Level  : {patient['risk']}")
            print(f"Health      : {patient['health']}%")


            found = True


    if found == False:

        print("Patient record not found.")



# ==========================================================
# UPDATE PATIENT
# ==========================================================


def update_patient(patient_list):

    update_id = input("Enter Patient ID to update : ").upper()


    for patient in patient_list:


        if patient["id"] == update_id:


            new_name = input("Enter New Name : ").title()


            patient["name"] = new_name


            print("Patient record updated successfully.")

            return



    print("Patient not found.")



# ==========================================================
# DELETE PATIENT
# ==========================================================


def delete_patient(patient_list):

    delete_id = input("Enter Patient ID to delete : ").upper()


    for patient in patient_list:


        if patient["id"] == delete_id:


            patient_list.remove(patient)


            print("Patient record deleted successfully.")

            return



    print("Patient not found.")



# ==========================================================
# VIEW ALL PATIENTS
# ==========================================================


def show_patients(patient_list):


    if len(patient_list) == 0:

        print("No patient records available.")

        return



    print("\nAll Patient Records")
    print("-" * 50)



    for patient in patient_list:


        print(f"ID : {patient['id']}")
        print(f"Name : {patient['name']}")
        print(f"Risk : {patient['risk']}")
        print("-" * 30)

# ==========================================================
#              FINAL PROJECT SUMMARY
# ==========================================================


print("\n")
print("=" * 65)
print("              MEDIRISK AI SUMMARY")
print("=" * 65)


print("Project Name : MediRisk AI")

print("System Type  : Patient Risk Assessment System")

print("Language     : Python")


print("-" * 65)


print("Features Available")

print("1. Patient Health Analysis")

print("2. BMI Calculation")

print("3. Risk Score Prediction")

print("4. Health Report Generation")

print("5. Patient Record Management")

print("6. Health Recommendations")


print("=" * 65)



# ==========================================================
#              PROGRAM END MESSAGE
# ==========================================================


print("\n")
print("=" * 65)

print("        Thank You For Using MediRisk AI")

print("   Intelligent Healthcare Risk Assessment System")

print("=" * 65)


print("\nStay Healthy! Take Care 😊")
```
